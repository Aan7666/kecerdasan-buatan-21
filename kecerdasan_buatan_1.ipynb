```python
# =========================
# IMPORT
# =========================

import tensorflow as tf
import numpy as np
import os

from tensorflow.keras import layers
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.preprocessing import image

from sklearn.metrics import confusion_matrix, classification_report

import seaborn as sns
import matplotlib.pyplot as plt

# =========================
# CEK GPU
# =========================

print("GPU:", tf.config.list_physical_devices('GPU'))

# =========================
# GOOGLE DRIVE
# =========================

from google.colab import drive
drive.mount('/content/drive')

# =========================
# PARAMETER
# =========================

img_size = (224,224)
batch_size = 32
# Sesuaikan jumlah kelas dengan dataset Pisang Muli kamu (Pisang Ambon, Muli, Raja, Tanduk)
num_classes = 4

# =========================
# DATASET PATH (COLAB)
# =========================

DATASET_PATH = '/content/drive/MyDrive/Citra_Pisang_Muli/jenis jenis pisang'

# =========================
# LOAD DATASET (DENGAN SPLIT 80/20)
# =========================

# Karena dataset kamu ada di 1 folder, kita pakai validation_split
train_ds = tf.keras.utils.image_dataset_from_directory(
    DATASET_PATH,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=img_size,
    batch_size=batch_size
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    DATASET_PATH,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=img_size,
    batch_size=batch_size
)

class_names = train_ds.class_names
print("Kelas:", class_names)

# =========================
# CACHE + PREFETCH (lebih cepat)
# =========================

AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.cache().prefetch(AUTOTUNE)
val_ds = val_ds.cache().prefetch(AUTOTUNE)
# Kita akan gunakan val_ds sebagai ganti test_ds untuk evaluasi akhir

# =========================
# DATA AUGMENTATION
# =========================

augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.2),
    layers.RandomZoom(0.2),
    layers.RandomContrast(0.2),
    layers.RandomBrightness(0.2),
    layers.RandomTranslation(0.1, 0.1)
])

# =========================
# CNN : MobileNetV2
# =========================

base_model = MobileNetV2(
    input_shape=(224,224,3),
    include_top=False,
    weights='imagenet'
)

# Fine tuning
base_model.trainable = True

for layer in base_model.layers[:-50]:
    layer.trainable = False

# =========================
# INPUT
# =========================

inputs = tf.keras.Input(shape=(224,224,3))

x = augmentation(inputs)
x = tf.keras.applications.mobilenet_v2.preprocess_input(x)
x = base_model(x)

# =========================
# CCT / TRANSFORMER BLOCK
# =========================

x = layers.Reshape((-1,x.shape[-1]))(x)

# positional embedding
seq_len = x.shape[1]

positions = tf.range(
    start=0,
    limit=seq_len,
    delta=1
)

position_embedding = layers.Embedding(
    input_dim=seq_len,
    output_dim=x.shape[-1]
)

embedded_positions = position_embedding(positions)
x = x + embedded_positions

# =========================
# TRANSFORMER ENCODER
# =========================

attention_output = layers.MultiHeadAttention(
    num_heads=4,
    key_dim=64,
    dropout=0.1
)(x,x)

x1 = layers.Add()([x,attention_output])
x1 = layers.LayerNormalization()(x1)

ffn = tf.keras.Sequential([
    layers.Dense(256, activation='relu'),
    layers.Dropout(0.2),
    layers.Dense(x.shape[-1])
])

ffn_output = ffn(x1)
x2 = layers.Add()([x1,ffn_output])
x2 = layers.LayerNormalization()(x2)

# =========================
# CLASSIFIER
# =========================

x = layers.GlobalAveragePooling1D()(x2)
x = layers.Dense(128, activation='relu')(x)
x = layers.Dropout(0.3)(x)

outputs = layers.Dense(
    num_classes,
    activation='softmax'
)(x)

# =========================
# MODEL
# =========================

model = tf.keras.Model(inputs, outputs)
model.summary()

# =========================
# COMPILE
# =========================

model.compile(
    optimizer=Adam(learning_rate=5e-5),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# =========================
# CALLBACKS
# =========================

early_stop = tf.keras.callbacks.EarlyStopping(
    monitor='val_accuracy',
    patience=5,
    restore_best_weights=True
)

reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.2,
    patience=2,
    verbose=1
)

# =========================
# TRAINING
# =========================

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=30,
    callbacks=[early_stop, reduce_lr]
)

# =========================
# TEST EVALUATION
# =========================

# Menggunakan val_ds karena dataset tidak dipisah ke folder test khusus
test_loss, test_acc = model.evaluate(val_ds)
print("\nEVALUATION ACCURACY (on validation set):", test_acc)

# =========================
# SAVE MODEL KE DRIVE
# =========================

SAVE_PATH = "/content/drive/MyDrive/model_cnn_cct_pisang.keras"
model.save(SAVE_PATH)

print("\nMODEL TERSIMPAN:")
print(SAVE_PATH)

# =========================
# CONFUSION MATRIX
# =========================

y_true = []
y_pred = []

for images, labels in val_ds:
    predictions = model.predict(images, verbose=0)
    predicted_labels = np.argmax(predictions, axis=1)
    y_true.extend(labels.numpy())
    y_pred.extend(predicted_labels)

cm = confusion_matrix(y_true, y_pred)

print("\nCONFUSION MATRIX:\n")
print(cm)

print("\nCLASSIFICATION REPORT:\n")
print(
    classification_report(
        y_true,
        y_pred,
        target_names=class_names,
        zero_division=0
    )
)

# =========================
# VISUALISASI
# =========================

plt.figure(figsize=(8,6))
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='magma',
    xticklabels=class_names,
    yticklabels=class_names
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("CNN + CCT Confusion Matrix")
plt.show()

# =========================
# PREDIKSI 1 GAMBAR
# =========================

IMAGE_PATH = '/content/drive/MyDrive/Citra_Pisang_Muli/jenis jenis pisang/pisang muli/Salinan pisang_muli_01.jpg'

img = image.load_img(
    IMAGE_PATH,
    target_size=(224,224)
)

img_array = image.img_to_array(img)
img_array = np.expand_dims(img_array, axis=0)

# prediksi
prediction = model.predict(img_array)
predicted_class = np.argmax(prediction)
confidence = np.max(prediction)

# =========================
# HASIL UTAMA
# =========================

print("\nHASIL PREDIKSI:")
print(f"Jenis Pisang: {class_names[predicted_class]}")
print(f"Confidence: {confidence*100:.2f}%")

# =========================
# DETAIL SEMUA PROBABILITAS
# =========================

print("\nSEMUA PROBABILITAS:\n")
for i, score in enumerate(prediction[0]):
    print(f"{class_names[i]} : {score*100:.2f}%")

# =========================
# TAMPILKAN GAMBAR
# =========================

plt.figure(figsize=(6,6))
plt.imshow(img)
plt.title(f"Prediksi: {class_names[predicted_class]}")
plt.axis('off')
plt.show()
```
